```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np


```

```python
# Paths to text files
ground_truth_file_path = '../recordings/reference_sessiontranskript.txt'
file_paths = ['./speaker-diarization-3-1-gpu/reference_session/8Speaker/seans3transkript.txt', './speaker-diarization-3-1-gpu/reference_session/12speaker/seans3transkript.txt', './medkit-simsamu-diarization/seans3transkript.txt', './whisper-timestamped/seans3transkript.txt', './reference_session-Speaker 0.txt']
```

```python
def read_file(file_path):
    with open(file_path, 'r', encoding='utf-8') as file:
        return file.read()

def calculate_similarity(ground_truth_path, file_paths):
    ground_truth_text = read_file(ground_truth_path)
    texts = [ground_truth_text] + [read_file(path) for path in file_paths]

    vectorizer = TfidfVectorizer()
    tfidf_matrix = vectorizer.fit_transform(texts)
    
    similarities = []
    for i in range(1, len(texts)):
        similarity = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[i:i+1])
        similarities.append(np.round(similarity[0][0], 4))

    return similarities
```

```python

# Calculating similarities
similarity_scores = calculate_similarity(ground_truth_file_path, file_paths)
for i, score in enumerate(similarity_scores):
    print(f"Similarity with file {file_paths[i]}: {score}")
```
